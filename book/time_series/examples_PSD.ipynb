# PSD examples
The following figure shows the autocorrelations and Power Spectral Densities (PSDs) for different signals.

In the first row, we observe the theoretical autocorrelation of white noise, which is a delta function $\delta (\tau)$ located at the zero lag with amplitude equal to the noise power $\sigma^2$. This reflects that white noise is uncorrelated, meaning its autocorrelation is zero for all non-zero lags. In reality, when computing the autocorrelation of a realization of white noise, we typically see a higher value of autocorrelation at lag zero, and values close to zero at other lags. The corresponding PSD of white noise is constant across all frequencies, indicating equal power distribution over the frequency spectrum.

In the second row, we observe a constant signal over time with value $A$. The autocorrelation of this signal is also a constant with a value of $A^2$ at all lags. The fact that the autocorrelation is flat indicates that a constant signal is perfecly correlated with itself at any time shift. Every sample has an identical value, so shifting the signal in time does not change the overlap. The corresponding PSD is a delta signal located at the zero frequency, indicating that the signal is not varying in time.

In the third row, an AR(1) process is shown.
